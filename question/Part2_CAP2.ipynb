## **PARTE 2- CONCEITOS ABORDADOS EM CADA QUIZ :**

**1. INTRODUÇÃO**

Apresenta a biblioteca Transformers da Hugging Face, destacando sua facilidade de uso, flexibilidade e simplicidade. A biblioteca permite carregar, treinar e salvar modelos Transformer com uma única API, facilitando o uso de modelos de última geração em PLN.

**2. POR TRÁS DO PIPELINE**

Explora o funcionamento interno da função pipeline(), detalhando as etapas de pré-processamento com o tokenizador, passagem dos inputs pelo modelo e pós-processamento. Demonstra como usar o AutoTokenizer para converter texto em tensores que podem ser processados pelo modelo.

**3. MODELOS**

Apresenta a classe AutoModel e suas variantes, que permitem instanciar modelos a partir de checkpoints. Mostra como criar um modelo Transformer, carregar configurações e pesos pré-treinados, e a importância de reutilizar modelos treinados para evitar esforços duplicados.

**4. TOKENIZADORES**

Explora os diferentes tipos de tokenizadores, como baseados em palavras, caracteres e subpalavras. Discute como os tokenizadores convertem texto em IDs numéricos, a importância de minimizar tokens desconhecidos e como diferentes estratégias de tokenização afetam o desempenho do modelo.

**5. LIDANDO COM MÚLTIPLAS SEQUÊNCIAS**

Aborda como os modelos Transformers esperam entradas em lotes (batches) e como lidar com múltiplas sequências de diferentes comprimentos. Explica o uso de padding, truncamento e máscaras de atenção para garantir que todas as sequências em um lote tenham o mesmo comprimento e sejam processadas corretamente pelo modelo.

**6. JUNTANDO TUDO**

Demonstra como o método tokenizer() pode simplificar o processo de tokenização, lidando automaticamente com padding, truncamento e conversão para tensores. Mostra como usar o tokenizador para processar sequências individuais ou múltiplas, preparando-as para serem passadas ao modelo.

**7. USO BÁSICO CONCLUÍDO**

Recapitula os conceitos aprendidos no capítulo, incluindo os componentes básicos de um modelo Transformer, o pipeline de tokenização e como usar modelos na prática. Encoraja os alunos a explorarem a documentação da Transformers, agora que estão familiarizados com a terminologia e os métodos principais.

**8. IMPLANTAÇÃO OTIMIZADA DE INFERÊNCIA**

Apresenta três frameworks modernos para servir LLMs com alto desempenho e eficiência em produção:

- Text Generation Inference (TGI) : Desenvolvido pela Hugging Face, usa Flash Attention 2 e batching contínuo para maximizar o uso da GPU. Ideal para ambientes de produção com estabilidade e escalabilidade.

- vLLM : Utiliza o PagedAttention, permitindo melhor gerenciamento de memória e cache compartilhado. Recomendado para múltiplas requisições simultâneas e uso eficiente da GPU.

- llama.cpp : Focado em execução local, com suporte a quantização (ex: 4 bits) para rodar modelos em CPUs. Ideal para dispositivos com poucos recursos, como notebooks e desktops.

Cada ferramenta atende a um cenário específico: TGI para produção robusta, vLLM para eficiência com múltiplos usuários e llama.cpp para uso local leve.
