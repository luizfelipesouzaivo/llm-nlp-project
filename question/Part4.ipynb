### **Parte 4- Engenharia de Prompts**

**Questão 1: Analise os seguintes prompts e identifique por que eles poderiam gerar respostas insatisfatórias ou irrelevantes:**

1. A)
**Exemplo 1: "Escreva sobre cachorros."**

O problema é que o tema é muito amplo. "Cachorros" pode incluir várias coisas, como raças, cuidados, comportamento... Sem um foco claro, a resposta pode ser vaga e sem direção.

1. B)
**Exemplo 2: "Explique física."**

Fica difícil explicar toda a física de uma vez, porque é um campo enorme. Sem especificar, a resposta pode ser genérica e não atender ao que você precisa.



**SUBQUESTÕES:**


**1. Reformule cada prompt utilizando técnicas de engenharia de prompts para torná-los mais específicos e direcionados.**

**EXEMPLO 1:** "Escreva um artigo sobre os cuidados essenciais para cachorros de raças pequenas, incluindo alimentação, higiene e prevenção de doenças comuns. O público-alvo são donos de cachorros iniciantes."


**EXEMPLO 2:** "Explique de maneira simples as leis de Newton, usando exemplos do dia a dia, como o movimento de um carro ou a queda de uma maçã. O objetivo é ajudar estudantes do ensino médio a entender esses conceitos de forma acessível e prática."


**2. Explique as melhorias feitas em cada caso e os motivos por trás das reformulações.**

**EXEMPLO 1:**

*MELHORIAS :*
1. Foco no tipo de cachorro (pequenos) e público-alvo (donos iniciantes).
2. Tópicos específicos como alimentação e cuidados de saúde.


*MOTIVOS :*

Tornar o conteúdo mais útil e fácil de entender para quem está começando a cuidar de um cachorro, sem sobrecarregar com informações desnecessárias.


**EXEMPLO 2:**

*MELHORIAS :*
1. Foco nas leis de Newton e exemplos simples do cotidiano.
2. Público-alvo específico (estudantes do ensino médio).


*MOTIVOS :*

Facilitar o entendimento para estudantes, usando exemplos práticos e simples que conectam teoria com situações do dia a dia.

**Questão 5: O prompt "Descreva a história da internet." foi mal formulado. Aplique técnicas de engenharia de prompts para melhorá-lo. Reformule o prompt para melhorar a especificidade e a qualidade da resposta. Justifique as mudanças feitas e explique como elas contribuem para obter uma resposta mais eficaz e relevante.**

**REFORMULAÇÃO :**

"Explique a evolução da internet, começando com a criação da ARPANET na década de 1960 até os dias de hoje. Inclua marcos importantes como o surgimento da World Wide Web e o impacto da internet no nosso dia a dia. A resposta deve ser simples e acessível para quem não tem conhecimento técnico."


**JUSTIFICATIVA :**

1. Especificação de tempo e marcos: Ao mencionar a ARPANET e a World Wide Web, o prompt direciona a resposta para eventos chave da história, garantindo que a resposta seja focada nos momentos mais importantes e não algo genérico. Isso melhora a organização e clareza.


2. Público-alvo claro: Pedir uma linguagem simples e acessível assegura que qualquer pessoa possa entender.


3. Impacto no cotidiano: Falar sobre como a internet afeta o nosso dia a dia deixa a resposta mais próxima da nossa realidade, tornando o assunto mais interessante e fácil de se conectar.


**CONTRIBUIÇÕES PARA UMA RESPOSTA EFICAZ :**


1. O novo prompt é mais claro e dá um foco específico, evitando que a resposta seja vaga ou ampla demais.


2. Ao focar nos marcos históricos e na evolução, a resposta será mais estruturada e fácil de entender, com uma linha do tempo lógica.


3. A linguagem acessível garante que o conteúdo seja engajador e compreensível para todos, sem perder profundidade.


**Questão 6: Aplique a técnica de Chain of Thought (CoT) para melhorar o prompt "Explique como funciona a energia solar.", detalhando o raciocínio necessário para que o modelo forneça uma resposta completa e coerente. Explique como a aplicação da técnica CoT melhora a resposta do modelo.**

"Explique como funciona a energia solar, começando com o que é a energia solar. Depois, conte como os painéis solares capturam a luz do sol e transformam em eletricidade, mencionando os principais componentes, como os painéis, inversores e baterias. Por fim, explique como a energia solar chega às casas ou empresas."

**Como a técnica CoT ajuda:**

1. Estrutura Clara: O modelo segue uma sequência lógica, começando do básico e indo até a distribuição da energia.

2. Fácil de Entender: Ao dividir a explicação em etapas, fica mais simples e direto, facilitando a compreensão.

3. Cobertura Completa: Garante que todos os pontos essenciais (captura, conversão, componentes e distribuição) sejam abordados de forma organizada.

No fim, o CoT torna a resposta mais fluida, sem perder detalhes importantes, com uma explicação clara e bem estruturada.
