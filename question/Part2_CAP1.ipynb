## **PARTE 1- CONCEITOS ABORDADOS EM CADA QUIZ :**

**1. INTRODUÇÃO**

Apresenta o curso e seus objetivos, destacando o uso das bibliotecas do ecossistema Hugging Face, como Transformers, Datasets, Tokenizers e Accelerate. Explica a diferença entre Processamento de Linguagem Natural (PLN) e Modelos de Linguagem de Grande Escala (LLMs), enfatizando a evolução do curso para focar em LLMs devido aos avanços na área.

**2. PROCESSAMENTO DE LINGUAGUEM NATURAL E LLMs**

Explora o que é PLN, incluindo tarefas comuns como classificação de texto, análise de sentimentos, tradução e geração de texto. Introduz os LLMs, destacando sua capacidade de realizar múltiplas tarefas sem treinamento específico, graças ao aprendizado em grandes volumes de dados.

**3. O QUE OS TRANSFORMERS PODEM FAZER?**

Apresenta a função pipeline() da biblioteca Transformers, que simplifica o uso de modelos pré-treinados para tarefas como análise de sentimentos. Mostra como os Transformers são aplicáveis em diversas modalidades, incluindo texto, imagens e áudio.

**4. COMO OS TRANSFORMERS FUNCIONAM?**

Detalha a arquitetura dos Transformers, introduzindo conceitos como mecanismos de atenção, codificadores e decodificadores. Discute o treinamento de modelos de linguagem e a evolução de modelos como GPT, BERT e T5.

**5. COMO OS TRANSFORMERS RESOLVEM TAREFAS**

Explica como diferentes arquiteturas de Transformers (codificador, decodificador e codificador-decodificador) são aplicadas a tarefas específicas, como classificação de texto, geração de texto e tradução. Apresenta exemplos de modelos como BERT, GPT2 e BART.

**6. ARQUITETURAS DE TRANSFORMERS**

Aprofunda-se nas três principais variantes arquitetônicas dos Transformers: apenas codificador, apenas decodificador e codificador-decodificador. Discute quando usar cada uma, com exemplos de modelos como BERT (codificador), GPT (decodificador) e T5 (codificador-decodificador).

**7. INFERÊNCIA COM LLMs**

Explora como os LLMs geram texto, enfatizando o papel do mecanismo de atenção e o conceito de comprimento de contexto. Discute estratégias de prompting para guiar a geração de texto de forma eficaz.

**8. VIÉS E LIMITAÇÕES**

Aborda os vieses presentes nos LLMs devido aos dados de treinamento, que podem refletir preconceitos sociais. Discute limitações como alucinações (geração de informações incorretas) e a falta de compreensão real do mundo pelos modelos.
