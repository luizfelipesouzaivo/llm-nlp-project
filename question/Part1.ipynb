**A) Pre-Training**

É a primeira fase de treinamento dos modelos de linguagem, como o GPT.
Nessa fase, o modelo aprende os padrões gerais da linguagem humana com base em grandes volumes de texto, geralmente sem rótulos.

**Explicação do Diagrama**

1. Corpus de Texto (Grande volume de dados):

O modelo é exposto a bilhões de palavras coletadas de livros, sites, artigos, fóruns, etc.
Exemplo: Wikipédia, StackOverflow, livros técnicos, jornais.

2. Tokenização :

O texto bruto é dividido em “tokens”, que podem ser palavras, pedaços de palavras ou até caracteres.
Exemplo: “computador” → ["com", "puta", "dor"]

3. Modelo LLM (Rede Neural Transformer) :

A arquitetura usada é a Transformer, que processa os tokens com camadas de atenção para entender relações entre palavras, mesmo que distantes.

4. Objetivo: Prever a Próxima Palavra :

O modelo é treinado para prever a próxima palavra em uma sequência.
Exemplo: "O céu está ___" → modelo aprende que "azul" é uma boa resposta.

**B) Transfer Learning**

Transfer learning em LLMs consiste em aproveitar um modelo previamente treinado em grandes volumes de dados (pré-treinamento) e adaptá-lo para tarefas específicas com poucos dados adicionais. Esse processo reduz o custo e o tempo de treinamento. O modelo reutiliza o conhecimento linguístico aprendido para entender melhor o novo domínio. É a base para aplicações eficientes em classificação, geração de texto e muito mais.

**Explicação do Diagrama :**

1. Pré-Treinamento Geral :

O modelo é treinado com um corpus massivo e genérico (ex: Wikipedia, livros, sites).

Objetivo: aprender padrões gerais da linguagem


2. Modelo Base (LLM) :

Após o pré-treinamento, temos um modelo "base" que compreende bem a estrutura da linguagem.
Exemplos: BERT, GPT, T5, RoBERTa.

3. Fine-Tuning :

O modelo base é ajustado com um corpus específico da tarefa (como perguntas e respostas, classificação de sentimentos, etc.).
Exemplo: Afinar o modelo com dados de e-mails para detectar spam.

4. Inferência :

O modelo treinado agora executa tarefas reais com mais precisão porque foi exposto a dados relevantes e específicos.

**C) Embeddings**

São representações numéricas densas de palavras, geradas para que modelos compreendam relações semânticas. Palavras com significados parecidos ficam próximas em um espaço vetorial. LLMs como o BERT convertem tokens em vetores de alta dimensão (ex: 768 valores). Isso permite que o modelo "entenda" similaridade e contexto.

**EXEMPLO:**

image.png

**Explicação do Diagrama :**

1. Palavras ou Tokens :

As palavras do texto (como “amor”, “inteligência”) são entradas simbólicas.

2. Embeddings :

Cada palavra é convertida em um vetor de números (ex: [0.12, -0.5, 0.88, ...]), que representa propriedades semânticas e sintáticas.

3. Espaço Vetorial :

Esses vetores são organizados em um espaço de alta dimensão onde palavras com significados semelhantes ficam próximas.
Exemplo: "rei" está próximo de "rainha", e distante de "banana".

**D) Transformers**

São a arquitetura base dos LLMs, compostos por blocos de atenção e redes neurais feedforward. Usam self-attention para processar todos os tokens em paralelo, capturando relações contextuais longas. Não dependem de recorrência, o que os torna altamente eficientes. São usados em modelos como BERT, GPT, T5 e similares.

**EXEMPLO:**

image.png

**Explicação do Diagrama :**

1. Input Embeddings :

As palavras são transformadas em vetores (embeddings) para serem processadas pela rede.

2. Positional Encoding :

Como Transformers não usam recorrência, adiciona-se codificações posicionais para indicar a ordem das palavras.

3. Self-Attention + Feed Forward Network

O encoder analisa todos os tokens da entrada simultaneamente, aplicando autoatenção para capturar dependências contextuais.

4. Self-Attention + Cross-Attention + FFN

O decoder prevê as saídas passo a passo, combinando atenção interna e atenção cruzada com os encoders.

5. Saída

Converte os vetores finais em probabilidades para prever o próximo token ou classe.

**E) Attention**

Permite que modelos foquem nas partes mais relevantes de uma frase. Cada palavra "olha" para outras usando vetores de consulta (Q), chave (K) e valor (V). O produto Q·K gera pesos que determinam o quanto cada palavra influencia na outra. Isso gera uma saída contextualizada essencial para o funcionamento dos Transformers.

**EXEMPLO :**

image.png

**Explicação do Diagrama**

1. Entrada(Tokens)

Cada palavra (token) é transformada em vetores para análise.

2. Q, K, V (Query, Key, Value)

Para cada token, o modelo gera três vetores:

* Q (pergunta sobre o contexto),

* K (chave para comparar),

* V (informação a ser agregada).

3. Cálculo de Similaridade (Q·Kᵀ)

O modelo calcula quanto cada token deve "prestar atenção" aos outros, usando o produto escalar entre Q e K.

4. Softmax → Pesos de Atenção

Os valores são normalizados para representar a importância relativa de cada token.

5. Pesos × V → Saída

Os pesos obtidos são usados para ponderar os valores V, gerando uma saída rica em contexto.

**F) Fine-tuning**

Processo de adaptar um LLM pré-treinado a uma tarefa específica, usando dados adicionais rotulados. O modelo ajusta seus pesos com base nesses dados, aprendendo a resolver melhor a tarefa final. É comum em aplicações como classificação de texto, análise de sentimentos ou perguntas e respostas. Esse processo torna o modelo mais útil e especializado.

**Explicação do Diagrama:**

1. Modelo Pré-Treinado

Um LLM como BERT ou GPT já foi treinado com grande volume de dados genéricos.

2. Corpus Específico

Fornecemos dados rotulados de um domínio específico (jurídico, médico, e-mails, etc.).

3. Treinamento Supervisionado

Ajustamos os pesos do modelo com base nos exemplos fornecidos, para melhorar o desempenho na nova tarefa.

4. Modelo Especializado

O resultado é um modelo adaptado e otimizado para a tarefa desejada (ex: análise de sentimentos, respostas jurídicas).
