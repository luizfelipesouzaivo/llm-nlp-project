**2. Explique a arquitetura do aplicativo, incluindo como o Streamlit, LLM e LangChain são utilizados.**

1. Streamlit

A camada de frontend da aplicação é construída com Streamlit, que permite criar rapidamente interfaces web usando apenas Python. A aplicação possui:

* Uma área de texto para o usuário digitar sua pergunta médica;

* Um menu de sugestões rápidas com dúvidas comuns;

* Um botão "Enviar" que dispara o processamento;

* Um painel de respostas com histórico das interações.

Além disso, a interface possui personalização visual por meio de CSS embutido (st.markdown(style)), garantindo uma aparência amigável e moderna.

2. LangChain

A camada de integração e orquestração da inteligência artificial é feita com a biblioteca LangChain, que atua como ponte entre o Streamlit e o modelo de linguagem (LLM).

* O módulo agent.py define a classe Agent, que encapsula o comportamento do assistente virtual.

* O LangChain utiliza a ChatOpenAI como chat_model, instanciado com uma OPENAI_API_KEY.

* A função get_tips() constrói um prompt personalizado com a pergunta do usuário e envia ao LLM via chat_model.invoke(prompt).

3. LLM

O modelo LLM usado é um modelo de conversa (ChatGPT) da OpenAI, acessado através da LangChain. Esse modelo é responsável por:

* Interpretar a pergunta do usuário;

* Responder de forma clara, empática e técnica, como se fosse um assistente médico virtual;

* Gerar respostas personalizadas, baseadas em linguagem natural e com alta qualidade semântica.

 1. agent.py

```python
from langchain_community.chat_models import ChatOpenAI

class Agent:
    def __init__(self):
        self.chat_model = ChatOpenAI(
            openai_api_key="OPENAI_API_KEY",
            temperature=0.6
        )

    def get_tips(self, pergunta):
        prompt = f"""
Você é um assistente médico virtual. Responda de forma clara, empática e profissional à seguinte dúvida médica:

Pergunta: {pergunta}
"""
        resposta = self.chat_model.invoke(prompt)
        return resposta.content
```

2. app.py

```python
import streamlit as st
from agent import Agent

st.set_page_config(page_title="Assistente Médico Virtual", layout="centered")

st.markdown("""
<style>
    .main { background-color: #eaf0fc; }
    .titulo { text-align: center; font-size: 28px; font-weight: bold; color: #007BFF; margin-top: 20px; }
    .subtitulo { text-align: center; font-size: 16px; margin-bottom: 30px; }
    .response-box {
        background-color: #d9ecff; padding: 15px; border-radius: 10px; margin-top: 20px;
    }
    .historico {
        font-size: 14px; color: #333;
        background-color: #f8f8f8; border-left: 5px solid #007BFF;
        padding: 10px; margin: 10px 0;
    }
    .stButton>button {
        background-color: #00BFFF; color: white;
        padding: 12px 24px; border-radius: 6px; font-size: 16px;
    }
</style>
""", unsafe_allow_html=True)

st.markdown('<div class="titulo"> Assistente Médico Virtual</div>', unsafe_allow_html=True)
st.markdown('<div class="subtitulo">Digite sua dúvida médica e receba uma resposta segura e clara.</div>', unsafe_allow_html=True)

sugestoes = [
    "Estou com dor de cabeça, o que posso tomar?",
    "Estou com febre há dois dias, o que devo fazer?",
    "Tenho tosse seca e dor no corpo, é covid?",
    "Quais são os sintomas de dengue?",
    "Tive tontura hoje cedo, o que pode ser?"
]

st.selectbox("Sugestões de sintomas:", [""] + sugestoes, key="sugestao", on_change=lambda: st.session_state.update(
    {"pergunta": st.session_state.sugestao}
))

pergunta = st.text_area("Digite sua pergunta médica:", key="pergunta", height=120)

if "historico" not in st.session_state:
    st.session_state.historico = []

if st.button("Enviar") and pergunta.strip():
    agent = Agent()
    resposta = agent.get_tips(pergunta)
    st.markdown(f'<div class="response-box"><b>Resposta:</b> {resposta}</div>', unsafe_allow_html=True)
    st.session_state.historico.append((pergunta, resposta))

if st.session_state.historico:
    st.markdown("### 🧾 Histórico de perguntas:")
    for i, (perg, resp) in enumerate(reversed(st.session_state.historico), 1):
        st.markdown(f"<div class='historico'><b>{i}. Pergunta:</b> {perg}<br><b>Resposta:</b> {resp}</div>", unsafe_allow_html=True)

st.markdown("""<br><hr><div style="text-align: center; font-size: 12px; color: gray;">
As respostas são geradas por IA e têm caráter informativo. Consulte sempre um médico de verdade para diagnósticos.
</div>""", unsafe_allow_html=True)
```
